## Linear Regression with Multiple Data Points

We are given the following dataset:

| x | y |
|---|---|
| 1 | 3 |
| 2 | 5 |
| 3 | 7 |
| 4 | 9 |

We want to find the best-fit straight line using the linear regression model:

$$\hat{y} = wx + b$$

where:

- $x$ = input
- $y$ = actual output
- $\hat{y}$ = predicted output
- $w$ = weight or slope
- $b$ = bias or intercept

### Initial values

Start with:

$$w = 0$$

$$b = 0$$

Use learning rate:

$$0.01$$

Run gradient descent for:

$$10 \text{ epochs}$$

### Step 1: Prediction

For each data point, calculate:

$$\hat{y}_i = wx_i + b$$

### Step 2: Error

Calculate the prediction error:

$$e_i = \hat{y}_i - y_i$$

### Step 3: Mean Squared Error

Calculate the Mean Squared Error:

$$MSE = \frac{1}{n}\sum_{i=1}^{n}(\hat{y}_i-y_i)^2$$

where:

$$n = 4$$

### Step 4: Calculate gradients

Gradient with respect to $w$:

$$\frac{\partial MSE}{\partial w} = \frac{2}{n} \sum_{i=1}^{n} (\hat{y}_i-y_i)x_i$$

Gradient with respect to $b$:

$$\frac{\partial MSE}{\partial b} = \frac{2}{n} \sum_{i=1}^{n} (\hat{y}_i-y_i)$$

### Step 5: Update parameters

Update the weight:

$$w_{\text{new}} = w_{\text{old}} - 0.01 \frac{\partial MSE}{\partial w}$$

Update the bias:

$$b_{\text{new}} = b_{\text{old}} - 0.01 \frac{\partial MSE}{\partial b}$$

### Task

For each epoch, print:

- Epoch number
- Current weight $w$
- Current bias $b$
- MSE loss
- Gradient with respect to $w$
- Gradient with respect to $b$

Finally, plot:

- The original data points
- The regression line before training
- The regression line after training

### Goal

Observe how gradient descent gradually adjusts $w$ and $b$ so that the regression line fits the data points better and the MSE becomes smaller.

## **1.Raw data points visualize**

```python
import matplotlib.pyplot as plt

# Dataset
x = [1, 2, 3, 4]
y = [3, 5, 7, 9]

# Scatter plot
plt.figure(figsize=(8, 5))

plt.scatter(x, y, s=100)

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Training Data")

plt.grid(True)

plt.show()
```

# **2. Training Linear Regression**

```python
import numpy as np
import matplotlib.pyplot as plt

def neuron(w, b, x):
    return w * x + b

def main():

    # Dataset
    x = np.array([1, 2, 3, 4], dtype=float)
    y = np.array([3, 5, 7, 9], dtype=float)

    # Initial w, b
    w = 0.0
    b = 0.0

    lr = 0.01
    epoch = 1000

    # Training
    for i in range(epoch):

        # Prediction
        yp = neuron(w, b, x)

        # Gradient
        dldw = np.mean((yp - y) * x)
        dldb = np.mean(yp - y)

        # Loss
        loss = np.mean(0.5 * (yp - y)**2)

        # Update
        w = w - lr * dldw
        b = b - lr * dldb

        if i % 10 == 0:
            print(
                f"Epoch: {i}, "
                f"w = {w:.4f}, "
                f"b = {b:.4f}, "
                f"Loss = {loss:.4f}"
            )

    # ---------------------------
    # Final prediction line
    # ---------------------------

    x_line = np.linspace(0, 5, 100)

    y_line = neuron(w, b, x_line)

    # ---------------------------
    # Plot
    # ---------------------------

    plt.figure(figsize=(8, 5))

    # Original data points
    plt.scatter(
        x,
        y,
        s=100,
        label="Actual Data"
    )

    # Learned regression line
    plt.plot(
        x_line,
        y_line,
        label=f"Learned Line: y = {w:.2f}x + {b:.2f}"
    )

    plt.xlabel("x")
    plt.ylabel("y")

    plt.title("Linear Regression using Gradient Descent")

    plt.grid(True)
    plt.legend()

    plt.show()


main()
```
