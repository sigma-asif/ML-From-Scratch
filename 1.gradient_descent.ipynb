## Gradient Descent Practice

Consider the one-variable function:

$$
f(x) = (x - 3)^2 + 2
$$

### Task

1. Find the derivative:

$$
f'(x)
$$

2. Start gradient descent from:

$$
x_0 = 8
$$

3. Use learning rate:

$$
0.1
$$

4. Run gradient descent for **10 epochs** using the update rule:

$$
x_{\text{new}} = x_{\text{old}} - 0.1 \times f'(x)
$$

5. For each epoch, print:

- Epoch number
- Current value of \(x\)
- Function value \(f(x)\)
- Gradient \(f'(x)\)

6. Plot:

- The function \(f(x)\)
- All gradient descent points for 10 epochs
- Connect the gradient descent points so that the movement toward the minimum can be seen clearly

### Goal

Observe how gradient descent gradually moves \(x\) toward the minimum point of the function.

```python
import numpy as np
import matplotlib.pyplot as plt

# Function
def f(x):
    return (x - 3)**2 + 2

# Derivative
def df(x):
    return 2 * (x - 3)

# Initial value
x = 8.0

# Settings
learning_rate = 0.1
epochs = 10

# Store values for plotting
x_history = []
y_history = []

# Gradient Descent
for epoch in range(1, epochs + 1):

    y = f(x)
    gradient = df(x)

    x_history.append(x)
    y_history.append(y)

    print(
        f"Epoch {epoch}: "
        f"x = {x:.4f}, "
        f"f(x) = {y:.4f}, "
        f"gradient = {gradient:.4f}"
    )

    # Update x
    x = x - 0.1 * gradient


# Plot the function
x_values = np.linspace(0, 9, 400)
y_values = f(x_values)

plt.figure(figsize=(9, 6))

plt.plot(x_values, y_values, label="f(x) = (x - 3)^2 + 2")

# Plot gradient descent points
plt.scatter(
    x_history,
    y_history,
    s=80,
    label="Gradient Descent Steps"
)

# Connect the points
plt.plot(
    x_history,
    y_history,
    linestyle="--"
)

# Mark the minimum point
plt.scatter(
    3,
    f(3),
    s=100,
    marker="x",
    label="Minimum"
)

plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Gradient Descent - 10 Epochs")
plt.grid(True)
plt.legend()

plt.show()
```

## Gradient Descent with Two Variables

Consider the function:

$$
f(x,y) = (x - 3)^2 + (y + 2)^2 + 1
$$

### Task

1. Find the partial derivative with respect to \(x\):

$$
\frac{\partial f}{\partial x}
$$

2. Find the partial derivative with respect to \(y\):

$$
\frac{\partial f}{\partial y}
$$

3. Write the gradient vector:

$$
\nabla f(x,y)
=
\begin{bmatrix}
\frac{\partial f}{\partial x} \\
\frac{\partial f}{\partial y}
\end{bmatrix}
$$

4. Start gradient descent from:

$$
x_0 = 8
$$

$$
y_0 = 5
$$

5. Use learning rate:

$$
0.1
$$

6. Run gradient descent for **10 epochs** using:

$$
x_{\text{new}}
=
x_{\text{old}}
-
0.1
\frac{\partial f}{\partial x}
$$

and

$$
y_{\text{new}}
=
y_{\text{old}}
-
0.1
\frac{\partial f}{\partial y}
$$

7. For each epoch, print:

- Epoch number
- Current \(x\)
- Current \(y\)
- Function value \(f(x,y)\)
- Gradient with respect to \(x\)
- Gradient with respect to \(y\)

8. Plot:

- The 3D surface of \(f(x,y)\)
- The gradient descent path for all 10 epochs
- The minimum point

### Goal

Observe how gradient descent updates both \(x\) and \(y\) and gradually moves toward the minimum point of the function.

```python
import numpy as np
import plotly.graph_objects as go

# Function
def f(x, y):
    return (x - 3)**2 + (y + 2)**2 + 1

# Partial derivatives
def dfdx(x, y):
    return 2 * (x - 3)

def dfdy(x, y):
    return 2 * (y + 2)

# Starting point
x = 8.0
y = 5.0

learning_rate = 0.1
epochs = 10

# Store history
x_history = []
y_history = []
z_history = []

# Gradient Descent
for epoch in range(1, epochs + 1):

    z = f(x, y)

    grad_x = dfdx(x, y)
    grad_y = dfdy(x, y)

    x_history.append(x)
    y_history.append(y)
    z_history.append(z)

    print(
        f"Epoch {epoch:2d} | "
        f"x = {x:.4f} | "
        f"y = {y:.4f} | "
        f"f(x,y) = {z:.4f} | "
        f"df/dx = {grad_x:.4f} | "
        f"df/dy = {grad_y:.4f}"
    )

    # Update
    x = x - learning_rate * grad_x
    y = y - learning_rate * grad_y


# Final point
x_history.append(x)
y_history.append(y)
z_history.append(f(x, y))


# Surface values
x_values = np.linspace(-2, 10, 100)
y_values = np.linspace(-8, 8, 100)

X, Y = np.meshgrid(x_values, y_values)
Z = f(X, Y)


# Create interactive figure
fig = go.Figure()


# 3D Surface
fig.add_trace(
    go.Surface(
        x=X,
        y=Y,
        z=Z,
        opacity=0.7,
        name="Function Surface"
    )
)


# Gradient Descent Path
fig.add_trace(
    go.Scatter3d(
        x=x_history,
        y=y_history,
        z=z_history,
        mode="lines+markers",
        marker=dict(size=5),
        line=dict(width=6),
        name="Gradient Descent"
    )
)


# Starting point
fig.add_trace(
    go.Scatter3d(
        x=[x_history[0]],
        y=[y_history[0]],
        z=[z_history[0]],
        mode="markers",
        marker=dict(size=8),
        name="Start (8, 5)"
    )
)


# Minimum point
fig.add_trace(
    go.Scatter3d(
        x=[3],
        y=[-2],
        z=[1],
        mode="markers",
        marker=dict(size=8, symbol="diamond"),
        name="Minimum (3, -2)"
    )
)


fig.update_layout(
    title="Interactive Gradient Descent",

    scene=dict(
        xaxis_title="x",
        yaxis_title="y",
        zaxis_title="f(x,y)"
    ),

    width=900,
    height=700
)


fig.show()
```
